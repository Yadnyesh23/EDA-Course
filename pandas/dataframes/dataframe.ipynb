# DataFrame

```python
import numpy as np
import pandas as pd
```

### Creating Dataframes

```python
# using list
student_list = [
    [80,95, 99],
    [9,9.5, 9.9],
    [8,10, 14]
]
pd.DataFrame(student_list, columns=['iq', 'cgpa', 'package'])
```

```python
# using dictionary
students_dict = {
    'name' : ['ABC', 'PQR', 'XYZ'],
    'iq' : [80,95, 99],
    'cgpa' : [9,9.5, 9.9],
    'package' :  [8,10, 14]
}

students_df = pd.DataFrame(students_dict)
students_df.set_index('name', inplace=True)
students_df
```
