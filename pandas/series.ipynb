# Series

### Series from List

```python
# string
country = ["India", "Brazil", "Russia", "China", "South Africa"]
pd.Series(country)
```

```python
# integer
runs = [12,23,34,45,67]
runs_series=pd.Series(runs)
```

```python
# custom index
languages = ["Python", "CPP", "C", "Javascript"]
pd.Series(languages, index=["a","b","c","d"])
```

```python
# custom index
marks = [97, 98, 96]
subjects = ["Physics", "Maths", "Chemistry"]
pd.Series(marks, index=subjects)
```

```python
# setting a name
pd.Series(marks, index=subjects, name="Yadnyesh ke marks")
```

### Series from dictionary

```python
marks = {
    "Physics" : 97,
    "Mathematics" : 98,
    "Chemistry" : 96
}

marks_series = pd.Series(marks, name="Yadnyesh ke marks")
```

### Series using read csv


```python
khl_run = pd.read_csv('pandas\kohli_ipl.csv', index_col='match_no').squeeze('runs')
khl_run
```

```python
bollywood = pd.read_csv('pandas\bollywood.csv', index_col='movie', sqeeze=True)
bollywood
```

### Series Methods

```python
# Head and Tail
bollywood.head() # by default it provides first 5
bollywood.head(7)  # it provides first 7

bollywood.tail()   #  by default it provides last 5
bollywood.tail(6)  # it provides last 6
```

```python
# Sample
bollywood.sample() # by default randomly provides 1 row
bollywood.sample(5) # randomly provides 5 rows
```

```python
# Value counts 
bollywood.value_counts()
```

```python
# Sort Values
bollywood.sort_values() # By default it sorts in ascending order of values
bollywood.sort_values(ascending=False) # This sorts in descending order
bollywood.sort_values(inplace=True) # Changes the original series
```

```python
# Sort Index
bollywood.sort_index() # By default it sorts in ascending order of index
bollywood.sort_index(ascending=False) # This sorts in descending order
bollywood.sort_index(inplace=True) # Changes the original series
```

### Series maths methods

```python
# sum and product
khl_run.sum()
khl_run.product()
```

```python
# mean, median, mode, std, var
khl_run.mean()
khl_run.median()
khl_run.mode()
khl_run.std()
khl_run.var()
```

```python
# min/max
khl_run.min()
khl_run.max()
```

```python
# describe
khl_run.describe
```

### Series indexing

```python
# integer indexing
x = pd.Series([1,2,3,4,5])
x[0]
```

```python
# negative indexing does not work in series
```

```python
# slicing
bollywood[5:16]
```

```python
# negative slicing
bollywood[-5:]
```

```python
# fancy indexing
# give me data of 1st, 3rd, 5th and 6th row
bollywood[[1,3,5,6]]
```

### Editing in series

```python
# using index
marks_series['Mathematics'] = 100
marks_series
```

```python
# if an index does not exist
marks_series['english'] = 98
marks_series
```

```python
# slicing
runs_series[0:3] = [100,100,100]
runs_series
```

```python
# fancy indexing
runs_series[[0,3,4]] = [20,20,20]
runs_series
```

### Series with python functionalities

```python
# len/type/dir/sorted/max/min
print(len(khl_run))
print(type(khl_run))
print(dir(khl_run))
print(sorted(khl_run))
print(max(khl_run))
print(min(khl_run))
```

```python
# type conversion
list(khl_run)
dict(khl_run)
```

```python
# membership operator - by default it searches in indexes
'Akshay Kumar' in bollywood
'Alia Bhatt' in bollywood.values
```
