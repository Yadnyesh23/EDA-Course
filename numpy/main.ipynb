# Indexing

### Indexing in 1D array

```python
arr
```

```python
print(arr[3]) # 3rd element
```

```python
arr[1:4] # from 1st to 3rd element
```

```python
arr[:]  # Last element
```

```python
arr[2:]  # from 2nd element till last
```

```python
arr[:4]  # From start till 4-1 = 3rd element
```

```python
print(arr[-1]) # Last one
```

```python
arr[:-1] # every on except last one
```

```python
arr[::] #return all element
```

```python
arr[::-1] # returns teh reverse of array
```

```python
arr[::-2] # return the array in reverse order but return element by skipping alternate
```

### Indexing in 2D array

```python
arr1
```

```python
arr1[:,:]  #   arr1[r, c] r = number of rows and c = number of columns
```

```python
arr1[:, 2]  # This means all the element from  2nd columns
```

```python
arr1[2,:]  #This means all elements of row at 2nd index
```

```python
# we want 4, 6 from 2nd row and 6, 9 from 3rd row
arr1[1:,1:3]
```

```python
# we want 8, 10 from 2nd row and 12, 15 from 3rd row
arr1[1:,3:]
```

```python
# we want 1,2,3 from 2nd row and 5, 10, 15 from 3rd row
arr1[:,::4]
```
